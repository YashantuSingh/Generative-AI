12 Extract Person Names
"Sachin Tendulkar lives in Mumbai"


```python
sentences = [
    "Sachin Tendulkar lives in Mumbai",
    "Virat Kohli and Rohit Sharma play for India",
    "Narendra Modi is the Prime Minister of India",
    "APJ Abdul Kalam was a great scientist",
]

for sentence in sentences:
    doc = nlp(sentence)
    names = [ent.text for ent in doc.ents if ent.label_ == "PERSON"]
    print(f"Text: {sentence}")
    print(f"Names: {names}\n")
```

22 Remove Multiple Emojis
"I love NLP ❤️😊🚀"

```python
import re

text = "I love NLP ❤️😊🚀"

emoji_pattern = re.compile(
    "["
    "\U0001F600-\U0001F64F"
    "\U0001F300-\U0001F5FF"
    "\U0001F680-\U0001F6FF"
    "\U0001F1E0-\U0001F1FF"
    "\U00002700-\U000027BF"
    "\U000024C2-\U0001F251"
    "]+",
    flags=re.UNICODE
)

clean_text = emoji_pattern.sub(r'', text)

print(clean_text)
```

25 Remove Stopwords
"This is a sample sentence"


```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk

nltk.download('punkt')
nltk.download('stopwords')

text = "This is a sample sentence"

words = word_tokenize(text)

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word.lower() not in stop_words]

print(filtered_words)
```

27 Basic Phrase Matching
"Machine learning is part of artificial intelligence"
"machine learning", "artificial intelligence"


```python
import spacy
from spacy.matcher import PhraseMatcher

nlp = spacy.load("en_core_web_sm")

doc = nlp("Machine learning is part of artificial intelligence")

matcher = PhraseMatcher(nlp.vocab)

phrases = ["machine learning", "artificial intelligence"]

patterns = [nlp(text) for text in phrases]

matcher.add("AI_TERMS", patterns)

matches = matcher(doc)

for match_id, start, end in matches:
    span = doc[start:end]
    print(span.text)
```

30 Count Phrase Matches
"Python is used in deep learning"
"deep learning", "python"


```python
import spacy
from spacy.matcher import PhraseMatcher

nlp = spacy.load("en_core_web_sm")

doc = nlp("Python is used in deep learning")

matcher = PhraseMatcher(nlp.vocab, attr="LOWER")

phrases = ["deep learning", "python"]

patterns = [nlp(text) for text in phrases]

matcher.add("TECH", patterns)

matches = matcher(doc)

print("Total Matches:", len(matches))

for match_id, start, end in matches:
    span = doc[start:end]
    print(span.text)
```
